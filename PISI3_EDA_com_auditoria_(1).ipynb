# 4 - Pré-processamento e Divisão dos Dados
(Tratamento de nulos, encoding, normalização e train_test_split.)


````python
features = ['animal_type', 'age_upon_intake_(days)', 'sex_upon_intake',
            'intake_condition', 'intake_type', 'intake_month',
            'intake_year', 'intake_weekday', 'intake_hour']

X = df[features].copy()
y = df['stay_category'].copy()

# Tratamento de nulos
# [ALTERADO] remove também o rótulo 'Desconhecido' (na base atual são 0 registros)
valid_mask = (~y.isna()) & (y != 'Desconhecido')
X, y = X[valid_mask], y[valid_mask]
X['age_upon_intake_(days)'] = X['age_upon_intake_(days)'].fillna(X['age_upon_intake_(days)'].median())
X['intake_weekday'] = X['intake_weekday'].fillna('Unknown')
X['intake_hour'] = X['intake_hour'].fillna(X['intake_hour'].median())

# Encoding
categorical_cols = ['animal_type', 'sex_upon_intake', 'intake_condition', 'intake_type', 'intake_weekday']
X_encoded_raw = pd.get_dummies(X, columns=categorical_cols, drop_first=True)
numeric_features = ['age_upon_intake_(days)', 'intake_month', 'intake_year', 'intake_hour']

# [ALTERADO - SEM VAZAMENTO] 1) divide primeiro...
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded_raw, y, test_size=0.2, random_state=42, stratify=y)
X_train, X_test = X_train.copy(), X_test.copy()

# ...2) e só depois padroniza, ajustando o scaler APENAS no treino
scaler = StandardScaler()
X_train[numeric_features] = scaler.fit_transform(X_train[numeric_features])
X_test[numeric_features] = scaler.transform(X_test[numeric_features])

# Base completa padronizada usada SOMENTE na clusterização (K-Means não usa rótulo nem teste)
X_encoded = X_encoded_raw.copy()
X_encoded[numeric_features] = StandardScaler().fit_transform(X_encoded[numeric_features])

print(f"Treino: {X_train.shape[0]} amostras | Teste: {X_test.shape[0]} amostras")
print("Distribuição do alvo no treino:", y_train.value_counts(normalize=True).round(3).to_dict())
````

# 5 - Clusterização (K-Means)

````python
# Limitando n_init para acelerar
inertias, silhouettes, K_range = [], [], range(2, 8) # Reduzi o range máximo para 8 para ser mais rápido

# Usamos uma amostra menor para calcular silhueta se o dataset for gigante
sample_size = min(10000, X_encoded.shape[0])
X_sample = X_encoded.sample(n=sample_size, random_state=42)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=5)
    kmeans.fit(X_encoded)
    inertias.append(kmeans.inertia_)
    # Silhueta calculada apenas na amostra para não travar
    silhouettes.append(silhouette_score(X_sample, kmeans.predict(X_sample)))

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].plot(K_range, inertias, marker='o'); axes[0].set_title('Cotovelo')
axes[1].plot(K_range, silhouettes, marker='o', color='green'); axes[1].set_title('Silhueta')
plt.show()

best_k = K_range[np.argmax(silhouettes)]
kmeans = KMeans(n_clusters=best_k, random_state=42, n_init=10)
df.loc[valid_mask, 'cluster'] = kmeans.fit_predict(X_encoded)
print(f"Melhor K encontrado: {best_k}")
````

````python
# [NOVO] Perfil de cada cluster com a proporção de permanência Prolongada (> 14 dias)
base_cluster = df.loc[valid_mask].copy()

perfil_clusters = base_cluster.groupby('cluster').agg(
    n_animais=('cluster', 'size'),
    idade_mediana_dias=('age_upon_intake_(days)', 'median'),
    especie_principal=('animal_type', lambda s: s.mode()[0]),
    especie_percentual=('animal_type', lambda s: s.value_counts(normalize=True).iloc[0] * 100),
    tipo_entrada_principal=('intake_type', lambda s: s.mode()[0]),
    permanencia_mediana_dias=('time_in_shelter_days', 'median'),
    permanencia_media_dias=('time_in_shelter_days', 'mean'),
    pct_longa=('stay_category', lambda s: (s == 'Prolongada').mean() * 100),
).round(2)

print(f"% de Prolongada na base inteira: {(base_cluster['stay_category'] == 'Prolongada').mean() * 100:.1f}%")
display(perfil_clusters)

pd.crosstab(base_cluster['cluster'], base_cluster['stay_category'], normalize='index').plot(
    kind='bar', stacked=True, figsize=(8, 4), color=['#4C9F70', '#D1495B'])
plt.title('Proporção de permanência Curta x Prolongada por cluster')
plt.ylabel('Proporção'); plt.xlabel('Cluster'); plt.xticks(rotation=0)
plt.legend(title='Permanência'); plt.tight_layout(); plt.show()
````

# 5.1 - [EXTRA] Checagem da análise por clusters

Diagnóstico da clusterização da seção 5, sem alterar o agrupamento já usado nos perfis:
1. se o rótulo de cluster entra como atributo dos modelos (risco de vazamento);
2. se o agrupamento depende do conjunto de teste (K-Means ajustado só no treino x base toda);
3. se os grupos são estáveis com outras sementes aleatórias;
4. se os clusters aparecem na mesma proporção no treino e no teste;
5. como os clusters se distribuem entre as classes **Curta** e **Prolongada** (qui-quadrado e V de Cramér).

````python
# =============================================================================
# 5.1 - [EXTRA] Checagem da análise por clusters (diagnóstico, não altera nada)
# Usa: df, valid_mask, X_encoded, X_train, X_test, y_train, y_test, kmeans, best_k
# Não muda o 'cluster' já gravado em df nem os modelos.
# =============================================================================
from scipy.stats import chi2_contingency
from sklearn.metrics import adjusted_rand_score

print("=" * 75)
print("CHECAGEM DA CLUSTERIZAÇÃO")
print("=" * 75)

# -----------------------------------------------------------------------------
# 1. Vazamento de dados: o cluster entrou como atributo do modelo?
# -----------------------------------------------------------------------------
colunas_cluster_no_modelo = [c for c in X_train.columns if 'cluster' in c.lower()]
print("\n1) O rótulo de cluster é atributo dos modelos supervisionados?")
print("   Colunas com 'cluster' em X_train:", colunas_cluster_no_modelo or "nenhuma")
print("   ->", "NÃO há vazamento pelo cluster: ele é usado só na análise descritiva (perfis)."
      if not colunas_cluster_no_modelo else "ATENÇÃO: o cluster entra no modelo; revise.")
print("   O K-Means não usa o alvo (stay_category) e foi ajustado em X_encoded,"
      " uma cópia separada da base, padronizada à parte.")

# -----------------------------------------------------------------------------
# 2. O agrupamento depende do conjunto de teste?
#    Ajusta o K-Means só no treino e compara com os rótulos atuais (base toda)
# -----------------------------------------------------------------------------
X_enc_treino = X_encoded.loc[X_train.index]
X_enc_teste = X_encoded.loc[X_test.index]

kmeans_treino = KMeans(n_clusters=best_k, random_state=42, n_init=10).fit(X_enc_treino)
rotulos_atuais_teste = df.loc[X_test.index, 'cluster'].astype(int)
rotulos_so_treino_teste = kmeans_treino.predict(X_enc_teste)
ari_treino_vs_base = adjusted_rand_score(rotulos_atuais_teste, rotulos_so_treino_teste)

print("\n2) K-Means ajustado SÓ no treino x K-Means atual (base toda), comparados no teste")
print(f"   Adjusted Rand Index (ARI) = {ari_treino_vs_base:.3f}  (1 = agrupamentos idênticos)")

# -----------------------------------------------------------------------------
# 3. Estabilidade: os grupos mudam com outra semente aleatória?
# -----------------------------------------------------------------------------
X_amostra_estab = X_encoded.sample(n=min(10000, len(X_encoded)), random_state=7)
rotulos_ref = kmeans.predict(X_amostra_estab)
aris_sementes = []
for semente in [0, 1, 2, 3, 4]:
    km = KMeans(n_clusters=best_k, random_state=semente, n_init=10).fit(X_encoded)
    aris_sementes.append(adjusted_rand_score(rotulos_ref, km.predict(X_amostra_estab)))
print("\n3) Estabilidade com 5 sementes diferentes (ARI em relação ao agrupamento atual)")
print("   ARI por semente:", [round(a, 3) for a in aris_sementes],
      f"| média = {np.mean(aris_sementes):.3f}")

# -----------------------------------------------------------------------------
# 4. Os clusters se distribuem igual entre treino e teste?
# -----------------------------------------------------------------------------
dist_split = pd.DataFrame({
    'Treino': df.loc[X_train.index, 'cluster'].value_counts(normalize=True).sort_index(),
    'Teste': df.loc[X_test.index, 'cluster'].value_counts(normalize=True).sort_index(),
}).mul(100).round(2)
dist_split.index = dist_split.index.astype(int)
dist_split.index.name = 'cluster'
print("\n4) Proporção de cada cluster no treino e no teste (%)")
display(dist_split)

# -----------------------------------------------------------------------------
# 5. Clusters x classes (Curta / Prolongada): há associação?
# -----------------------------------------------------------------------------
base_cluster = df.loc[valid_mask].copy()
tabela_cluster_classe = pd.crosstab(base_cluster['cluster'].astype(int), base_cluster['stay_category'])
chi2, p_valor, gl, _ = chi2_contingency(tabela_cluster_classe)
n_total = tabela_cluster_classe.values.sum()
cramers_v = np.sqrt(chi2 / (n_total * (min(tabela_cluster_classe.shape) - 1)))

print("\n5) Distribuição das classes de permanência dentro de cada cluster")
display(pd.concat([
    tabela_cluster_classe,
    pd.crosstab(base_cluster['cluster'].astype(int), base_cluster['stay_category'],
                normalize='index').mul(100).round(2).add_suffix(' (%)')
], axis=1))
print(f"   Qui-quadrado = {chi2:.1f} | gl = {gl} | p-valor = {p_valor:.2e}")
print(f"   V de Cramér = {cramers_v:.3f}  (0 = nenhuma associação; < 0,1 = associação desprezível)")

# -----------------------------------------------------------------------------
# 6. Resumo
# -----------------------------------------------------------------------------
checagem_clusters = pd.DataFrame([
    {'Verificação': 'Cluster usado como atributo do modelo', 'Resultado': 'Não' if not colunas_cluster_no_modelo else 'Sim'},
    {'Verificação': 'ARI K-Means só no treino x base toda (teste)', 'Resultado': round(ari_treino_vs_base, 3)},
    {'Verificação': 'ARI médio entre sementes', 'Resultado': round(float(np.mean(aris_sementes)), 3)},
    {'Verificação': 'Maior diferença de proporção treino x teste (p.p.)',
     'Resultado': round(float((dist_split['Treino'] - dist_split['Teste']).abs().max()), 2)},
    {'Verificação': 'V de Cramér (cluster x classe)', 'Resultado': round(float(cramers_v), 3)},
    {'Verificação': 'p-valor do qui-quadrado', 'Resultado': f"{p_valor:.2e}"},
])
print("\n" + "=" * 75)
print("RESUMO DA CHECAGEM")
print("=" * 75)
display(checagem_clusters)
````


# 6 - Treinamento dos Modelos Clássicos
(Smote isolado para termos uma visão dos resultados originais + prevenção de perda de memória.)

````python
models = {
    'Random Forest': RandomForestClassifier(n_estimators=50, random_state=42, n_jobs=-1),
    'Logistic Regression': LogisticRegression(max_iter=500, random_state=42, n_jobs=-1),
    'KNN': KNeighborsClassifier(n_jobs=-1)
    # 'SVM': SVC(probability=True, random_state=42) # <- DESCOMENTE POR SUA CONTA E RISCO
}

# [ALTERADO - BINÁRIO]
classes = ['Curta', 'Prolongada']

def extrair_metricas(y_true, y_pred):
    report = classification_report(y_true, y_pred, output_dict=True, zero_division=0)
    return {
        'Accuracy': accuracy_score(y_true, y_pred),
        'Macro F1': report['macro avg']['f1-score'],
        'F1 Prolongada': report['Prolongada']['f1-score'],
        'Recall Prolongada': report['Prolongada']['recall'],
        'Precisão Prolongada': report['Prolongada']['precision'],
    }

def avaliar(name, model, X_fit, y_fit, cenario):
    """Treina em (X_fit, y_fit) e mede no treino ORIGINAL e no teste (para ver overfitting)."""
    model.fit(X_fit, y_fit)
    tr = extrair_metricas(y_train, model.predict(X_train))
    te = extrair_metricas(y_test, model.predict(X_test))
    return {
        'Modelo': name, 'Balanceamento': cenario,
        'Train Accuracy': tr['Accuracy'], 'Train Macro F1': tr['Macro F1'],
        'Test Accuracy': te['Accuracy'], 'Test Macro F1': te['Macro F1'],
        'Test F1 Prolongada': te['F1 Prolongada'], 'Test Recall Prolongada': te['Recall Prolongada'],
        'Test Precisão Prolongada': te['Precisão Prolongada'],
        'Gap F1 (treino - teste)': tr['Macro F1'] - te['Macro F1'],
    }

results_no_smote = []
for name, model in models.items():
    print(f"Treinando {name}...")
    results_no_smote.append(avaliar(name, model, X_train, y_train, 'Sem SMOTE'))

print("Treinamento sem SMOTE concluído!")
pd.DataFrame(results_no_smote).round(4)
````

# 7.2 - [NOVO] Controle de overfitting da Random Forest

# 7.3 - [NOVO] Validação cruzada (5 folds estratificados, somente no treino)

````python
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.base import clone

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
candidatos_cv = {
    'Random Forest': RandomForestClassifier(n_estimators=50, random_state=42, n_jobs=-1),
    'Random Forest (class_weight)': RandomForestClassifier(n_estimators=50, random_state=42, n_jobs=-1, class_weight='balanced'),
    'Random Forest (regularizada)': clone(rf_regularizada),
    'Logistic Regression (class_weight)': LogisticRegression(max_iter=500, random_state=42, class_weight='balanced'),
    'KNN': KNeighborsClassifier(n_jobs=-1),
}

linhas_cv = []
for nome, modelo in candidatos_cv.items():
    print(f"Validação cruzada: {nome}...")
    scores = cross_val_score(modelo, X_train, y_train, cv=cv, scoring='f1_macro', n_jobs=-1)
    linhas_cv.append({'Modelo': nome, 'F1 macro (média CV)': scores.mean(), 'Desvio-padrão': scores.std()})

resultados_cv = pd.DataFrame(linhas_cv).sort_values('F1 macro (média CV)', ascending=False)
display(resultados_cv.round(4))
````

# 7.4 - [NOVO] Ajuste do limiar de decisão (modelo final)
Por padrão o modelo diz "Prolongada" quando a probabilidade passa de 0,5. Escolhemos o limiar que maximiza o F1 da classe Prolongada usando **previsões fora da amostra do próprio treino** (`cross_val_predict`). O teste continua intocado até a avaliação final.

````python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import f1_score, confusion_matrix, ConfusionMatrixDisplay

modelo_final = clone(rf_regularizada)

proba_oof = cross_val_predict(modelo_final, X_train, y_train, cv=cv, method='predict_proba', n_jobs=-1)
modelo_final.fit(X_train, y_train)
idx_longa = list(modelo_final.classes_).index('Prolongada')
proba_oof_longa = proba_oof[:, idx_longa]

limiares = np.round(np.arange(0.20, 0.71, 0.01), 2)
f1_por_limiar = [f1_score(y_train == 'Prolongada', proba_oof_longa >= t) for t in limiares]
limiar_otimo = float(limiares[int(np.argmax(f1_por_limiar))])

plt.figure(figsize=(7, 4))
plt.plot(limiares, f1_por_limiar, marker='.')
plt.axvline(limiar_otimo, color='red', linestyle='--', label=f'Limiar escolhido = {limiar_otimo:.2f}')
plt.axvline(0.5, color='gray', linestyle=':', label='Limiar padrão = 0,50')
plt.title('F1 da classe Prolongada por limiar (validação cruzada no treino)')
plt.xlabel('Limiar de probabilidade para "Prolongada"'); plt.ylabel('F1 Prolongada'); plt.legend(); plt.show()

proba_teste_longa = modelo_final.predict_proba(X_test)[:, idx_longa]
pred_padrao = np.where(proba_teste_longa >= 0.5, 'Prolongada', 'Curta')
pred_ajustada = np.where(proba_teste_longa >= limiar_otimo, 'Prolongada', 'Curta')

print(f"=== MODELO FINAL (RF regularizada) - limiar 0,50 ===")
print(classification_report(y_test, pred_padrao, digits=4, zero_division=0))
print(f"=== MODELO FINAL (RF regularizada) - limiar {limiar_otimo:.2f} ===")
print(classification_report(y_test, pred_ajustada, digits=4, zero_division=0))

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
for ax, pred, titulo in [(axes[0], pred_padrao, 'Limiar 0,50'), (axes[1], pred_ajustada, f'Limiar {limiar_otimo:.2f}')]:
    ConfusionMatrixDisplay(confusion_matrix(y_test, pred, labels=classes), display_labels=classes).plot(ax=ax, colorbar=False)
    ax.set_title(titulo)
plt.tight_layout(); plt.show()
````

# 7.4.1 - Auditoria de Equidade (Fairness) do Modelo Final

Esta etapa avalia se o modelo final apresenta diferenças de desempenho entre grupos de animais definidos por espécie, sexo e faixa etária.

A auditoria utiliza exclusivamente a Random Forest regularizada já treinada na seção 7.4, mantendo o mesmo conjunto de teste e o limiar de decisão selecionado por validação cruzada no conjunto de treino.

Serão comparadas métricas de classificação da classe Prolongada, incluindo quantidade de animais, precisão, recall, F1 e erros de classificação por grupo.

O objetivo é identificar possíveis disparidades de desempenho entre os grupos avaliados. A análise não altera o modelo, não modifica o limiar e não representa uma certificação de ausência de viés.

````python
# ============================================================
# AUDITORIA DE EQUIDADE — MODELO FINAL DO ADOTAI
# ============================================================
# Utiliza exclusivamente:
# modelo_final, X_test, y_test, limiar_otimo e df.
#
# Não retreina o modelo.
# Não altera o limiar.
# Não modifica o conjunto de teste.
# Não modifica o bundle do Dashboard.
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
    confusion_matrix
)

print("=" * 75)
print("AUDITORIA DE EQUIDADE — VALIDAÇÃO DO MODELO FINAL")
print("=" * 75)

# ------------------------------------------------------------
# 1. VALIDAR A EXISTÊNCIA DAS VARIÁVEIS
# ------------------------------------------------------------

necessarias = [
    "modelo_final",
    "X_test",
    "y_test",
    "limiar_otimo",
    "df",
    "features"
]

ausentes = [v for v in necessarias if v not in globals()]

if ausentes:
    raise NameError(
        f"Variáveis necessárias não encontradas: {ausentes}"
    )

features_esperadas = [
    'animal_type',
    'age_upon_intake_(days)',
    'sex_upon_intake',
    'intake_condition',
    'intake_type',
    'intake_month',
    'intake_year',
    'intake_weekday',
    'intake_hour'
]

if list(features) != features_esperadas:
    raise ValueError(
        "A lista de features não corresponde às nove variáveis "
        "originais do modelo final."
    )

if not isinstance(y_test, pd.Series):
    raise TypeError("y_test precisa ser uma Series do pandas.")

if not X_test.index.equals(y_test.index):
    raise ValueError(
        "Os índices de X_test e y_test não estão alinhados."
    )

if not df.index.is_unique:
    raise ValueError(
        "O índice original de df não é único. "
        "Não é seguro associar os grupos às amostras."
    )

if not y_test.index.isin(df.index).all():
    raise ValueError(
        "Há índices de teste que não foram encontrados em df."
    )

colunas_grupo = [
    'animal_type',
    'sex_upon_intake',
    'age_upon_intake_(days)'
]

colunas_ausentes = [
    c for c in colunas_grupo
    if c not in df.columns
]

if colunas_ausentes:
    raise KeyError(
        f"Colunas necessárias para a auditoria ausentes: "
        f"{colunas_ausentes}"
    )

if X_test.shape[1] != modelo_final.n_features_in_:
    raise ValueError(
        "O número de colunas de X_test não corresponde "
        "ao número de features utilizadas pelo modelo final."
    )

# ------------------------------------------------------------
# 2. IDENTIFICAR A CLASSE POSITIVA
# ------------------------------------------------------------

classes_modelo = list(modelo_final.classes_)

if 'Prolongada' not in classes_modelo:
    raise ValueError(
        "A classe Prolongada não foi encontrada no modelo."
    )

idx_prolongada = classes_modelo.index('Prolongada')

if set(y_test.dropna().unique()) != {'Curta', 'Prolongada'}:
    raise ValueError(
        "Os rótulos reais do teste não correspondem "
        "às classes Curta e Prolongada."
    )

# ------------------------------------------------------------
# 3. CALCULAR AS PREVISÕES DO MODELO FINAL
# ------------------------------------------------------------

# Não utiliza predict(), pois precisamos aplicar explicitamente
# o mesmo limiar de decisão utilizado pelo Dashboard.

probabilidades = modelo_final.predict_proba(X_test)[:, idx_prolongada]

y_real = y_test.astype(str).to_numpy()

previsoes = np.where(
    probabilidades >= limiar_otimo,
    'Prolongada',
    'Curta'
)

# ------------------------------------------------------------
# 4. VALIDAR AS MÉTRICAS GLOBAIS DO DASHBOARD
# ------------------------------------------------------------

metricas_globais = {
    'Acurácia': accuracy_score(
        y_real,
        previsoes
    ),
    'F1 macro': f1_score(
        y_real,
        previsoes,
        labels=['Curta', 'Prolongada'],
        average='macro',
        zero_division=0
    ),
    'Precisão Prolongada': precision_score(
        y_real,
        previsoes,
        pos_label='Prolongada',
        zero_division=0
    ),
    'Recall Prolongada': recall_score(
        y_real,
        previsoes,
        pos_label='Prolongada',
        zero_division=0
    ),
    'ROC-AUC': roc_auc_score(
        (y_real == 'Prolongada').astype(int),
        probabilidades
    )
}

referencias_dashboard = {
    'Acurácia': 0.722,
    'F1 macro': 0.666,
    'Precisão Prolongada': 0.450,
    'Recall Prolongada': 0.630,
    'ROC-AUC': 0.775
}

casas_dashboard = {
    'Acurácia': 3,
    'F1 macro': 3,
    'Precisão Prolongada': 2,
    'Recall Prolongada': 2,
    'ROC-AUC': 3
}

print("\nMÉTRICAS GLOBAIS OBTIDAS")
print("-" * 75)

for nome, valor in metricas_globais.items():
    referencia = referencias_dashboard[nome]

    print(
        f"{nome}: {valor:.4f} "
        f"| Referência Dashboard: {referencia:.{casas_dashboard[nome]}f}"
    )

print(f"\nAmostras de teste: {len(y_test):,}")
print(f"Limiar utilizado: {limiar_otimo:.4f}")
print(f"Classes do modelo: {classes_modelo}")

# As referências do Dashboard são exibidas com três casas decimais.
# A comparação utiliza a mesma precisão de apresentação.

validacao_metricas = {
    nome: round(valor, casas_dashboard[nome])
    == round(referencias_dashboard[nome], casas_dashboard[nome])
    for nome, valor in metricas_globais.items()
}

limiar_validado = round(float(limiar_otimo), 2) == 0.51

print("\nVALIDAÇÃO DAS REFERÊNCIAS")
print("-" * 75)

for nome, passou in validacao_metricas.items():
    print(f"{nome}: {'OK' if passou else 'DIVERGENTE'}")

print(
    f"Limiar 0,51: "
    f"{'OK' if limiar_validado else 'DIVERGENTE'}"
)

if len(y_test) != 15935:
    print(
        "\nATENÇÃO: a quantidade de amostras difere "
        "da referência do Dashboard (15.935)."
    )

if not all(validacao_metricas.values()) or not limiar_validado or len(y_test) != 15935:
    raise RuntimeError(
        "\nAUDITORIA INTERROMPIDA: as métricas globais, o limiar "
        "ou a quantidade de amostras não reproduziram as referências "
        "do Dashboard. Não prossiga para a análise por grupos. "
        "Investigue a divergência primeiro."
    )

print("\nVALIDAÇÃO GLOBAL APROVADA.")
print("O modelo final reproduziu as referências do Dashboard.")

# ------------------------------------------------------------
# 5. CONSTRUIR BASE DE AUDITORIA COM ÍNDICES ALINHADOS
# ------------------------------------------------------------

# Recupera as características originais dos mesmos animais
# que aparecem em X_test e y_test.

grupos_originais = df.loc[y_test.index, colunas_grupo].copy()

auditoria = pd.DataFrame(index=y_test.index)

auditoria['real'] = y_real
auditoria['previsto'] = previsoes
auditoria['probabilidade_prolongada'] = probabilidades

auditoria['especie'] = (
    grupos_originais['animal_type']
    .astype('string')
    .fillna('Não informado')
)

auditoria['sexo'] = (
    grupos_originais['sex_upon_intake']
    .astype('string')
    .fillna('Não informado')
)

# ------------------------------------------------------------
# 6. DEFINIR FAIXAS ETÁRIAS ANTES DA COMPARAÇÃO
# ------------------------------------------------------------

idade = pd.to_numeric(
    grupos_originais['age_upon_intake_(days)'],
    errors='coerce'
)

auditoria['faixa_etaria'] = pd.cut(
    idade,
    bins=[0, 365, 2555, np.inf],
    labels=[
        'Filhote (< 1 ano)',
        'Adulto (1 a < 7 anos)',
        'Idoso (>= 7 anos)'
    ],
    right=False
).astype('string').fillna('Não informado/idade inválida')

# ------------------------------------------------------------
# 7. FUNÇÃO DE MÉTRICAS POR GRUPO
# ------------------------------------------------------------

def calcular_metricas_grupo(dados, dimensao, categoria):

    y_true = (dados['real'] == 'Prolongada').astype(int)
    y_pred = (dados['previsto'] == 'Prolongada').astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    ).ravel()

    n = len(dados)
    suporte_prolongada = int(y_true.sum())
    suporte_curta = int(n - suporte_prolongada)

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    precisao = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    acuracia = accuracy_score(
        y_true,
        y_pred
    )

    taxa_falso_positivo = (
        fp / (fp + tn)
        if (fp + tn) > 0
        else np.nan
    )

    # Critério operacional de alerta:
    # grupos com menos de 30 animais ou menos de 10
    # exemplos de qualquer classe exigem cautela.
    amostra_limitada = (
        n < 30
        or suporte_prolongada < 10
        or suporte_curta < 10
    )

    return {
        'Dimensão': dimensao,
        'Grupo': str(categoria),
        'N': n,
        'N real Curta': suporte_curta,
        'N real Prolongada': suporte_prolongada,
        'Acurácia': acuracia,
        'Precisão Prolongada': precisao,
        'Recall Prolongada': recall,
        'F1 Prolongada': f1,
        'Taxa falso positivo': taxa_falso_positivo,
        'Falsos negativos': int(fn),
        'Falsos positivos': int(fp),
        'Amostra limitada': amostra_limitada
    }

# ------------------------------------------------------------
# 8. AUDITAR ESPÉCIE, SEXO E FAIXA ETÁRIA
# ------------------------------------------------------------

dimensoes = [
    'especie',
    'sexo',
    'faixa_etaria'
]

resultados_grupos = []

for dimensao in dimensoes:

    for categoria, grupo in auditoria.groupby(
        dimensao,
        dropna=False,
        observed=True
    ):

        resultados_grupos.append(
            calcular_metricas_grupo(
                grupo,
                dimensao,
                categoria
            )
        )

resultado_fairness = pd.DataFrame(resultados_grupos)

print("\n" + "=" * 75)
print("RESULTADOS DA AUDITORIA POR SUBGRUPOS")
print("=" * 75)

for dimensao in dimensoes:

    print(f"\nDIMENSÃO: {dimensao.upper()}")
    print("-" * 75)

    tabela = resultado_fairness[
        resultado_fairness['Dimensão'] == dimensao
    ].copy()

    display(
        tabela[
            [
                'Grupo',
                'N',
                'N real Curta',
                'N real Prolongada',
                'Acurácia',
                'Precisão Prolongada',
                'Recall Prolongada',
                'F1 Prolongada',
                'Taxa falso positivo',
                'Falsos negativos',
                'Amostra limitada'
            ]
        ].round(4)
    )

# ------------------------------------------------------------
# 9. RESUMO DESCRITIVO DAS DIFERENÇAS ENTRE GRUPOS
# ------------------------------------------------------------

resumo_disparidades = []

for dimensao in dimensoes:

    tabela = resultado_fairness[
        resultado_fairness['Dimensão'] == dimensao
    ].copy()

    # Somente grupos com quantidade suficiente de exemplos
    # das duas classes entram nesta comparação descritiva.

    grupos_com_suporte = tabela[
        (tabela['N'] >= 30)
        & (tabela['N real Curta'] >= 10)
        & (tabela['N real Prolongada'] >= 10)
    ]

    if len(grupos_com_suporte) >= 2:

        amplitude_recall = (
            grupos_com_suporte['Recall Prolongada'].max()
            - grupos_com_suporte['Recall Prolongada'].min()
        )

        amplitude_precisao = (
            grupos_com_suporte['Precisão Prolongada'].max()
            - grupos_com_suporte['Precisão Prolongada'].min()
        )

        grupo_menor_recall = grupos_com_suporte.loc[
            grupos_com_suporte['Recall Prolongada'].idxmin(),
            'Grupo'
        ]

        grupo_maior_recall = grupos_com_suporte.loc[
            grupos_com_suporte['Recall Prolongada'].idxmax(),
            'Grupo'
        ]

        resumo_disparidades.append({
            'Dimensão': dimensao,
            'Grupos comparáveis': len(grupos_com_suporte),
            'Menor recall': grupos_com_suporte['Recall Prolongada'].min(),
            'Maior recall': grupos_com_suporte['Recall Prolongada'].max(),
            'Diferença de recall': amplitude_recall,
            'Diferença de precisão': amplitude_precisao,
            'Grupo com menor recall': grupo_menor_recall,
            'Grupo com maior recall': grupo_maior_recall
        })

    else:

        resumo_disparidades.append({
            'Dimensão': dimensao,
            'Grupos comparáveis': len(grupos_com_suporte),
            'Menor recall': np.nan,
            'Maior recall': np.nan,
            'Diferença de recall': np.nan,
            'Diferença de precisão': np.nan,
            'Grupo com menor recall': 'Dados insuficientes',
            'Grupo com maior recall': 'Dados insuficientes'
        })

resumo_disparidades = pd.DataFrame(resumo_disparidades)

print("\n" + "=" * 75)
print("RESUMO DAS DIFERENÇAS ENTRE GRUPOS")
print("=" * 75)

display(resumo_disparidades.round(4))

# ------------------------------------------------------------
# 10. RESUMO DE CONTROLE
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("CONTROLE FINAL DA AUDITORIA")
print("=" * 75)

print(f"Modelo avaliado: {type(modelo_final).__name__}")
print(f"Total de amostras: {len(auditoria):,}")
print(f"Limiar aplicado: {limiar_otimo:.2f}")
print("Features originais:", features)
print("Dimensões auditadas:", dimensoes)

print("\nQuantidade de grupos com amostra limitada:")
print(
    resultado_fairness.groupby('Dimensão')['Amostra limitada']
    .sum()
    .to_dict()
)

print("\nAuditoria concluída.")
print(
    "Os resultados são comparações descritivas de desempenho "
    "por subgrupo e não constituem, isoladamente, uma certificação "
    "de ausência de viés."
)
````

### Interpretação e limitações da auditoria

A auditoria compara o desempenho do modelo final entre grupos de animais definidos por espécie, sexo e faixa etária, mantendo o mesmo conjunto de teste e o limiar de decisão utilizado pelo Dashboard.

As diferenças observadas nas métricas podem indicar que o modelo apresenta desempenho desigual entre determinados grupos. Entretanto, essas diferenças devem ser interpretadas considerando a quantidade de amostras, a distribuição das classes e a disponibilidade de exemplos em cada grupo.

Grupos com amostras pequenas ou baixa representação de uma das classes não permitem conclusões robustas. Por isso, seus resultados devem ser apresentados com ressalvas.

Esta análise representa uma avaliação de desempenho por subgrupos. Ela não comprova, isoladamente, a ausência ou a presença de discriminação, nem constitui uma certificação completa de equidade do modelo.

Os resultados serão utilizados para documentar a transparência do modelo e orientar a interpretação de seu desempenho no Dashboard.

````python
#============================================================
# AUDITORIA DE EQUIDADE — ETAPA 1
# Validação das métricas globais do modelo final
# Não retreina nem modifica o modelo.
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
    confusion_matrix
)

# 1. Verificações iniciais
assert 'modelo_final' in globals(), "Modelo final não encontrado."
assert 'X_test' in globals(), "X_test não encontrado."
assert 'y_test' in globals(), "y_test não encontrado."
assert 'limiar_otimo' in globals(), "Limiar não encontrado."

assert len(X_test) == len(y_test), "X_test e y_test têm tamanhos diferentes."
assert isinstance(y_test, pd.Series), "y_test não é uma Series."

# 2. Identificar corretamente a classe Prolongada
classes_modelo = list(modelo_final.classes_)

assert 'Prolongada' in classes_modelo, (
    "A classe Prolongada não foi encontrada no modelo."
)

idx_prolongada = classes_modelo.index('Prolongada')

# 3. Obter probabilidades e aplicar o limiar utilizado no Dashboard
probabilidades = modelo_final.predict_proba(X_test)[:, idx_prolongada]

previsoes_auditoria = np.where(
    probabilidades >= limiar_otimo,
    'Prolongada',
    'Curta'
)

# 4. Calcular métricas globais
y_real = y_test.to_numpy()

metricas_globais = {
    'Acurácia': accuracy_score(y_real, previsoes_auditoria),
    'F1 macro': f1_score(
        y_real, previsoes_auditoria,
        labels=['Curta', 'Prolongada'],
        average='macro',
        zero_division=0
    ),
    'Precisão Prolongada': precision_score(
        y_real, previsoes_auditoria,
        pos_label='Prolongada',
        zero_division=0
    ),
    'Recall Prolongada': recall_score(
        y_real, previsoes_auditoria,
        pos_label='Prolongada',
        zero_division=0
    ),
    'ROC-AUC': roc_auc_score(
        (y_real == 'Prolongada').astype(int),
        probabilidades
    )
}

print("===== VALIDAÇÃO DO MODELO FINAL =====")
print(f"Quantidade de exemplos: {len(y_test)}")
print(f"Limiar utilizado: {limiar_otimo:.4f}")
print(f"Classes do modelo: {classes_modelo}")

print("\n===== MÉTRICAS GLOBAIS =====")
for nome, valor in metricas_globais.items():
    print(f"{nome}: {valor:.4f}")

print("\n===== MATRIZ DE CONFUSÃO =====")
print(
    confusion_matrix(
        y_real,
        previsoes_auditoria,
        labels=['Curta', 'Prolongada']
    )
)
````

# 7.4.2 - [EXTRA] Auditoria de equidade: raças e fenótipos discriminados na adoção

Cria a coluna de auditoria **`Raca_Discriminada` ("Sim"/"Não")** a partir das colunas originais `breed` e `color`, **somente para os animais do conjunto de teste e depois da previsão**. A coluna não entra no modelo final: serve apenas para comparar o desempenho entre os dois grupos.

"Sim" quando o cão ou gato tem pelo menos um destes traços:
- **raça estigmatizada** (cães): Pit Bull, Staffordshire, American Bulldog, Rottweiler, Doberman, Chow Chow, Bull Terrier, Presa Canario, Cane Corso, Dogo Argentino;
- **pelagem preta** (cães e gatos): primeira cor registrada = Black;
- **"caramelo"** (cães): sem raça definida (Mix) com pelagem Tan, Fawn, Red, Buff, Gold, Yellow ou Apricot.

SRD (Mix) sozinho cobre cerca de 85% dos cães e gatos, por isso fica fora da regra principal e é analisado à parte e numa variante (`Raca_Discriminada_com_SRD`). Aves e "Other" ficam fora, pois esses fenótipos não se aplicam a eles.

````python
# =============================================================================
# 7.4.2 - [EXTRA] Auditoria de equidade: raças e fenótipos discriminados na adoção
# Atributo de AUDITORIA criado só sobre o conjunto de teste, depois da previsão.
# Não entra no modelo, não retreina, não muda o limiar nem o bundle do Dashboard.
# Usa: df, y_test, previsoes e probabilidades (da célula 7.4.1),
#      calcular_metricas_grupo (da célula 7.4.1)
# =============================================================================
import re

# Recalcula as previsões caso a célula 7.4.1 não tenha sido executada antes
if 'previsoes' not in globals() or len(previsoes) != len(y_test):
    probabilidades = modelo_final.predict_proba(X_test)[:, list(modelo_final.classes_).index('Prolongada')]
    previsoes = np.where(probabilidades >= limiar_otimo, 'Prolongada', 'Curta')

# Usa a função da auditoria 7.4.1; se ela não existir neste notebook, define uma igual
if 'calcular_metricas_grupo' not in globals():
    from sklearn.metrics import confusion_matrix, recall_score, precision_score, f1_score, accuracy_score
    def calcular_metricas_grupo(dados, dimensao, categoria):
        y_true = (dados['real'] == 'Prolongada').astype(int)
        y_pred = (dados['previsto'] == 'Prolongada').astype(int)
        tn, fp, fn, tp = confusion_matrix(y_true, y_pred, labels=[0, 1]).ravel()
        n, sup_p = len(dados), int(y_true.sum())
        return {'Dimensão': dimensao, 'Grupo': str(categoria), 'N': n,
                'N real Curta': n - sup_p, 'N real Prolongada': sup_p,
                'Acurácia': accuracy_score(y_true, y_pred),
                'Precisão Prolongada': precision_score(y_true, y_pred, zero_division=0),
                'Recall Prolongada': recall_score(y_true, y_pred, zero_division=0),
                'F1 Prolongada': f1_score(y_true, y_pred, zero_division=0),
                'Taxa falso positivo': fp / (fp + tn) if (fp + tn) else np.nan,
                'Falsos negativos': int(fn), 'Falsos positivos': int(fp),
                'Amostra limitada': n < 30 or sup_p < 10 or (n - sup_p) < 10}

# -----------------------------------------------------------------------------
# 1. Regras dos grupos (texto das colunas originais 'breed' e 'color')
# -----------------------------------------------------------------------------
# Raças de cães estigmatizadas (tipo "bully"/"perigosas"), com menor taxa de adoção na literatura
RACAS_ESTIGMATIZADAS = [
    'Pit Bull', 'Staffordshire', 'American Bulldog', 'Rottweiler', 'Doberman',
    'Chow Chow', 'Bull Terrier', 'Presa Canario', 'Cane Corso', 'Dogo Argentino',
]
# Cores que correspondem ao "caramelo" (vira-lata de pelagem caramelo/dourada)
CORES_CARAMELO = ['Tan', 'Fawn', 'Red', 'Buff', 'Gold', 'Yellow', 'Apricot']

padrao_estigma = re.compile('|'.join(RACAS_ESTIGMATIZADAS), flags=re.IGNORECASE)

originais = df.loc[y_test.index, ['animal_type', 'breed', 'color']].copy()
raca = originais['breed'].astype(str)
cor_principal = (originais['color'].astype(str)
                 .str.split('/').str[0]      # primeira cor (ex.: 'Black/White' -> 'Black')
                 .str.split(' ').str[0])     # sem o padrão (ex.: 'Brown Tabby' -> 'Brown')
eh_cao = originais['animal_type'].eq('Dog')
eh_gato = originais['animal_type'].eq('Cat')

auditoria_raca = pd.DataFrame(index=y_test.index)
auditoria_raca['real'] = y_test.astype(str).to_numpy()
auditoria_raca['previsto'] = previsoes
auditoria_raca['probabilidade_prolongada'] = probabilidades
auditoria_raca['especie'] = originais['animal_type']

auditoria_raca['is_raca_estigmatizada'] = (eh_cao & raca.str.contains(padrao_estigma)).astype(int)
auditoria_raca['is_pelagem_preta'] = ((eh_cao | eh_gato) & cor_principal.eq('Black')).astype(int)
auditoria_raca['is_srd'] = ((eh_cao | eh_gato) & raca.str.contains('Mix', case=False)).astype(int)
auditoria_raca['is_caramelo'] = (eh_cao & auditoria_raca['is_srd'].eq(1)
                                 & cor_principal.isin(CORES_CARAMELO)).astype(int)

# Coluna binária pedida: "Sim" se o animal tem ao menos um traço discriminado.
# SRD sozinho NÃO entra na regra principal (é ~85% dos cães e gatos); ele é analisado à parte
# e numa variante de sensibilidade logo abaixo.
tracos = ['is_raca_estigmatizada', 'is_pelagem_preta', 'is_caramelo']
auditoria_raca['Raca_Discriminada'] = np.where(auditoria_raca[tracos].any(axis=1), 'Sim', 'Não')
auditoria_raca['Raca_Discriminada_com_SRD'] = np.where(
    auditoria_raca[tracos + ['is_srd']].any(axis=1), 'Sim', 'Não')

# Só cães e gatos: os fenótipos acima não se aplicam a aves e "Other"
auditoria_raca = auditoria_raca[auditoria_raca['especie'].isin(['Dog', 'Cat'])]

print("=" * 75)
print("AUDITORIA DE EQUIDADE — RAÇAS E FENÓTIPOS DISCRIMINADOS (cães e gatos do teste)")
print("=" * 75)
print(f"Animais auditados: {len(auditoria_raca):,} de {len(y_test):,} do teste")
print("\nQuantos animais têm cada traço:")
display(auditoria_raca[tracos + ['is_srd']].sum().rename('N').to_frame()
        .assign(**{'% dos auditados': lambda t: (t['N'] / len(auditoria_raca) * 100).round(1)}))

# -----------------------------------------------------------------------------
# 2. Métricas por grupo (mesma função da auditoria 7.4.1)
# -----------------------------------------------------------------------------
def metricas_por_dimensao(dados, coluna, rotulo=None):
    linhas = [calcular_metricas_grupo(g, rotulo or coluna, cat)
              for cat, g in dados.groupby(coluna, observed=True)]
    tabela = pd.DataFrame(linhas)
    taxa_prevista = dados.groupby(coluna)['previsto'].apply(lambda s: (s == 'Prolongada').mean())
    taxa_real = dados.groupby(coluna)['real'].apply(lambda s: (s == 'Prolongada').mean())
    tabela['Taxa real Prolongada'] = tabela['Grupo'].map(taxa_real.rename(index=str))
    tabela['Taxa prevista Prolongada'] = tabela['Grupo'].map(taxa_prevista.rename(index=str))
    return tabela

colunas_exibir = ['Dimensão', 'Grupo', 'N', 'Taxa real Prolongada', 'Taxa prevista Prolongada',
                  'Precisão Prolongada', 'Recall Prolongada', 'F1 Prolongada',
                  'Taxa falso positivo', 'Falsos negativos', 'Amostra limitada']

resultado_raca = pd.concat([
    metricas_por_dimensao(auditoria_raca, 'Raca_Discriminada'),
    metricas_por_dimensao(auditoria_raca, 'Raca_Discriminada_com_SRD'),
    *[metricas_por_dimensao(auditoria_raca.assign(**{t: auditoria_raca[t].map({1: 'Sim', 0: 'Não'})}), t)
      for t in tracos + ['is_srd']],
], ignore_index=True)

print("\nMÉTRICAS DO MODELO FINAL POR GRUPO")
display(resultado_raca[colunas_exibir].round(4))

# Controle por espécie: cães têm recall menor que gatos no modelo final (auditoria 7.4.1),
# e as raças estigmatizadas e o caramelo só existem em cães. Comparar dentro da espécie
# separa o efeito da raça/fenótipo do efeito da espécie.
resultado_raca_especie = pd.concat([
    metricas_por_dimensao(auditoria_raca[auditoria_raca['especie'] == esp], 'Raca_Discriminada',
                          rotulo=f'Raca_Discriminada ({esp})')
    for esp in ['Dog', 'Cat']
], ignore_index=True)
print("\nMESMA COMPARAÇÃO DENTRO DE CADA ESPÉCIE")
display(resultado_raca_especie[colunas_exibir].round(4))

# -----------------------------------------------------------------------------
# 3. Diferenças "Sim" x "Não" com intervalo de confiança (bootstrap)
# -----------------------------------------------------------------------------
def diferencas_equidade(dados, coluna, n_boot=1000, semente=42):
    def calc(d):
        sim, nao = d[d[coluna] == 'Sim'], d[d[coluna] == 'Não']
        def taxas(g):
            real = g['real'].eq('Prolongada').to_numpy()
            prev = g['previsto'].eq('Prolongada').to_numpy()
            tpr = prev[real].mean() if real.any() else np.nan
            fpr = prev[~real].mean() if (~real).any() else np.nan
            return prev.mean(), tpr, fpr
        ps, rs, fs = taxas(sim)
        pn, rn, fn_ = taxas(nao)
        return {'Paridade demográfica (dif. taxa prevista)': ps - pn,
                'Razão de impacto (Sim/Não)': ps / pn if pn > 0 else np.nan,
                'Igualdade de oportunidade (dif. recall)': rs - rn,
                'Dif. taxa de falso positivo': fs - fn_}
    pontual = calc(dados)
    rng = np.random.default_rng(semente)
    valores = dados.reset_index(drop=True)
    boots = [calc(valores.iloc[rng.integers(0, len(valores), len(valores))]) for _ in range(n_boot)]
    boots = pd.DataFrame(boots)
    return pd.DataFrame({
        'Medida': list(pontual.keys()),
        'Valor': list(pontual.values()),
        'IC 95% inferior': boots.quantile(0.025).values,
        'IC 95% superior': boots.quantile(0.975).values,
    }).assign(Grupo_comparado=coluna)

diferencas_raca = pd.concat([
    diferencas_equidade(auditoria_raca, 'Raca_Discriminada'),
    diferencas_equidade(auditoria_raca, 'Raca_Discriminada_com_SRD'),
    diferencas_equidade(auditoria_raca[auditoria_raca['especie'] == 'Dog'], 'Raca_Discriminada').assign(Grupo_comparado='Raca_Discriminada (só cães)'),
    diferencas_equidade(auditoria_raca[auditoria_raca['especie'] == 'Cat'], 'Raca_Discriminada').assign(Grupo_comparado='Raca_Discriminada (só gatos)'),
], ignore_index=True)

print("\nDIFERENÇAS ENTRE 'SIM' E 'NÃO' (valor positivo = maior no grupo 'Sim')")
display(diferencas_raca.round(4))
print("Leitura: paridade demográfica perto de 0 e razão de impacto entre 0,8 e 1,25 indicam"
      " taxas de previsão parecidas; um IC que inclui 0 indica diferença não significativa.")

# -----------------------------------------------------------------------------
# 4. Gráfico: recall e taxa prevista por grupo
# -----------------------------------------------------------------------------
graf = resultado_raca[resultado_raca['Dimensão'] == 'Raca_Discriminada'].set_index('Grupo')
graf[['Taxa real Prolongada', 'Taxa prevista Prolongada', 'Recall Prolongada', 'Precisão Prolongada']].T.plot(
    kind='bar', figsize=(9, 4), color=['#4C9F70', '#D1495B'])
plt.title('Modelo final por grupo de raça/fenótipo discriminado (cães e gatos do teste)')
plt.ylabel('Proporção'); plt.xticks(rotation=0); plt.legend(title='Raca_Discriminada')
plt.tight_layout(); plt.show()

print("\nControle: modelo, limiar e conjunto de teste inalterados "
      f"(limiar = {limiar_otimo:.2f}, teste = {len(y_test):,}).")
````

# 7.7 - [EXTRA] Experimento comparativo: SMOTENC x SMOTE

O SMOTE da seção 7 foi aplicado depois do *one-hot encoding*. Ao interpolar dois vizinhos, ele gera valores entre 0 e 1 nas colunas categóricas, e o `imblearn` devolve essas colunas no tipo original (booleano): qualquer valor diferente de zero vira `True`. Com isso surgem exemplos sintéticos impossíveis, como um animal que é cão e gato ao mesmo tempo. O **SMOTENC** (*SMOTE for Nominal and Continuous*) interpola apenas os atributos numéricos e escolhe as categorias entre os vizinhos mais próximos, então todo exemplo sintético corresponde a um animal possível.

O experimento é aplicado **somente em `X_train`**, usa o mesmo conjunto de teste e a mesma função `avaliar` da seção 6 e treina **cópias** dos modelos. O modelo final (RF regularizada com `class_weight='balanced'` e limiar 0,51) não é alterado.

````python
# =============================================================================
# 7.7 - [EXTRA] SMOTENC x SMOTE (experimento comparativo, não altera o modelo final)
# Usa: X, X_train, X_test, y_train, y_test, scaler, numeric_features,
#      categorical_cols, models, rf_regularizada, avaliar, results_smote, results_reg
# =============================================================================
from imblearn.over_sampling import SMOTENC
from sklearn.base import clone

# -----------------------------------------------------------------------------
# 1. Por que SMOTENC: o SMOTE da seção 7 interpola as colunas one-hot. O imblearn
#    devolve essas colunas no tipo original (bool), então qualquer valor entre 0 e 1
#    vira True: aparecem animais "cão e gato ao mesmo tempo" ou com duas condições
#    de entrada. O SMOTENC interpola só as numéricas e sorteia as categóricas
#    entre os vizinhos, então todo exemplo sintético é um animal possível.
# -----------------------------------------------------------------------------
dummies = [c for c in X_train.columns if c not in numeric_features]

def linhas_impossiveis(X_sint):
    """Linhas com mais de uma categoria ativa no mesmo atributo (one-hot inválido)."""
    invalida = pd.Series(False, index=X_sint.index)
    for c in categorical_cols:
        grupo = [d for d in dummies if d.startswith(c + '_')]
        invalida |= (X_sint[grupo].astype(float) != 0).sum(axis=1) > 1
    return invalida

n_orig = len(X_train)
sinteticos_smote = X_train_res.iloc[n_orig:]   # o imblearn acrescenta os sintéticos no fim
inval_smote = linhas_impossiveis(sinteticos_smote)
print(f"SMOTE (seção 7): {inval_smote.sum():,} de {len(sinteticos_smote):,} exemplos sintéticos "
      f"({inval_smote.mean():.1%}) têm combinação impossível de categorias")
print("   ex.: cão e gato ao mesmo tempo:",
      int((sinteticos_smote['animal_type_Cat'].astype(bool) & sinteticos_smote['animal_type_Dog'].astype(bool)).sum()))

# -----------------------------------------------------------------------------
# 2. Base de treino ANTES do one-hot (mesmas linhas de X_train) e numéricas
#    padronizadas com o MESMO scaler da seção 4 (ajustado só no treino)
# -----------------------------------------------------------------------------
X_train_cat = X.loc[X_train.index, features].copy()
X_train_cat[numeric_features] = scaler.transform(X_train_cat[numeric_features])
for c in categorical_cols:
    X_train_cat[c] = X_train_cat[c].astype(str)

indices_categoricos = [X_train_cat.columns.get_loc(c) for c in categorical_cols]
print("Colunas categóricas informadas ao SMOTENC:", categorical_cols)
print("Colunas numéricas (interpoladas):", numeric_features)

smotenc = SMOTENC(categorical_features=indices_categoricos, random_state=42)
X_res_cat, y_train_smotenc = smotenc.fit_resample(X_train_cat, y_train)
print("Distribuição após SMOTENC:", pd.Series(y_train_smotenc).value_counts().to_dict())

# -----------------------------------------------------------------------------
# 3. One-hot igual ao da seção 4 e mesmas colunas de X_train
# -----------------------------------------------------------------------------
X_train_smotenc = (pd.get_dummies(X_res_cat, columns=categorical_cols, drop_first=True)
                   .reindex(columns=X_train.columns, fill_value=0)
                   .astype(X_train.dtypes.to_dict()))
assert list(X_train_smotenc.columns) == list(X_train.columns)
inval_smotenc = linhas_impossiveis(X_train_smotenc.iloc[n_orig:])
print(f"SMOTENC: {inval_smotenc.sum():,} de {len(X_train_smotenc) - n_orig:,} exemplos sintéticos "
      "com combinação impossível de categorias")

# -----------------------------------------------------------------------------
# 4. Treina CÓPIAS dos modelos (clone) para não alterar os objetos já usados
#    nas seções seguintes. Teste e métricas: mesma função 'avaliar' da seção 6.
# -----------------------------------------------------------------------------
results_smotenc = []
for name, model in models.items():
    print(f"Treinando {name} com SMOTENC...")
    results_smotenc.append(avaliar(name, clone(model), X_train_smotenc, y_train_smotenc, 'Com SMOTENC'))

# Configuração do modelo final (RF regularizada), sem class_weight porque os dados já estão balanceados
rf_reg_sem_peso = clone(rf_regularizada).set_params(class_weight=None)
results_rf_reg_over = [
    avaliar('Random Forest (regularizada)', clone(rf_reg_sem_peso), X_train_res, y_train_res, 'Com SMOTE'),
    avaliar('Random Forest (regularizada)', clone(rf_reg_sem_peso), X_train_smotenc, y_train_smotenc, 'Com SMOTENC'),
]

# -----------------------------------------------------------------------------
# 5. Comparação
# -----------------------------------------------------------------------------
colunas_cmp = ['Modelo', 'Balanceamento', 'Test Accuracy', 'Test Macro F1',
               'Test F1 Prolongada', 'Test Recall Prolongada', 'Test Precisão Prolongada',
               'Gap F1 (treino - teste)']
comparacao_smotenc = (pd.DataFrame(results_smote + results_smotenc + results_rf_reg_over + results_reg)
                      [colunas_cmp].sort_values(['Modelo', 'Balanceamento']).reset_index(drop=True))
print("\nSMOTE x SMOTENC (mesmo conjunto de teste; limiar 0,50)")
display(comparacao_smotenc.round(4))

ganho = (pd.DataFrame(results_smotenc).set_index('Modelo')[['Test Accuracy', 'Test Macro F1', 'Test F1 Prolongada']]
         - pd.DataFrame(results_smote).set_index('Modelo')[['Test Accuracy', 'Test Macro F1', 'Test F1 Prolongada']])
print("\nDiferença SMOTENC - SMOTE (positivo = SMOTENC melhor)")
display(ganho.round(4))

pivot = comparacao_smotenc.pivot_table(index='Modelo', columns='Balanceamento', values='Test Macro F1')
pivot.plot(kind='bar', figsize=(9, 4), rot=0)
plt.title('F1 macro no teste: SMOTE x SMOTENC'); plt.ylabel('F1 macro'); plt.ylim(0.4, 0.75)
plt.legend(title='Balanceamento'); plt.tight_layout(); plt.show()

print("Modelo final (seção 7.4) não foi alterado: continua a RF regularizada com class_weight "
      f"e limiar {limiar_otimo:.2f}.")
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.utils.class_weight import compute_sample_weight
from xgboost import XGBClassifier
from sklearn.metrics import classification_report

# 1. Atualizar a lista de features com as novas variáveis criadas
features_com_novas = [
    'animal_type', 'age_upon_intake_(days)', 'sex_upon_intake',
    'intake_condition', 'intake_type', 'intake_month',
    'intake_year', 'intake_weekday', 'intake_hour',
    'is_mix', 'is_black_or_dark', 'simplified_breed'
]

X_novo = df[features_com_novas].copy()
y_novo = df['stay_category'].copy()

# Tratamento de nulos das novas variáveis e antigas
valid_mask_novo = (~y_novo.isna()) & (y_novo != 'Desconhecido')
X_novo, y_novo = X_novo[valid_mask_novo], y_novo[valid_mask_novo]
X_novo['age_upon_intake_(days)'] = X_novo['age_upon_intake_(days)'].fillna(X_novo['age_upon_intake_(days)'].median())
X_novo['intake_weekday'] = X_novo['intake_weekday'].fillna('Unknown')
X_novo['intake_hour'] = X_novo['intake_hour'].fillna(X_novo['intake_hour'].median())
X_novo['is_mix'] = X_novo['is_mix'].fillna(0)
X_novo['is_black_or_dark'] = X_novo['is_black_or_dark'].fillna(0)
X_novo['simplified_breed'] = X_novo['simplified_breed'].fillna('Outros')

# Encoding e Scaling considerando a nova coluna categórica 'simplified_breed'
categorical_cols_novas = ['animal_type', 'sex_upon_intake', 'intake_condition', 'intake_type', 'intake_weekday', 'simplified_breed']
X_encoded_novo = pd.get_dummies(X_novo, columns=categorical_cols_novas, drop_first=True)

numeric_features_novas = ['age_upon_intake_(days)', 'intake_month', 'intake_year', 'intake_hour']

# [ALTERADO - SEM VAZAMENTO] divide primeiro e padroniza depois, com o scaler ajustado só no treino
X_train_n, X_test_n, y_train_n, y_test_n = train_test_split(
    X_encoded_novo, y_novo, test_size=0.2, random_state=42, stratify=y_novo
)
X_train_n, X_test_n = X_train_n.copy(), X_test_n.copy()
scaler_novo = StandardScaler()
X_train_n[numeric_features_novas] = scaler_novo.fit_transform(X_train_n[numeric_features_novas])
X_test_n[numeric_features_novas] = scaler_novo.transform(X_test_n[numeric_features_novas])

# 2. Encoding do Alvo (Target)
le_n = LabelEncoder()
y_train_num_n = le_n.fit_transform(y_train_n)
y_test_num_n = le_n.transform(y_test_n)
class_names_n = [str(cls) for cls in le_n.classes_]

# 3. Pesos para compensar o desbalanceamento
sample_weights_n = compute_sample_weight(class_weight='balanced', y=y_train_num_n)

# 4. Treinar XGBoost Classifier
xgb_novo_modelo = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=6,
    random_state=42,
    n_jobs=-1,
    eval_metric='logloss'  # [ALTERADO] binário
)

print("Treinando XGBoost Weighted com as Novas Features de Raça e Cor...")
xgb_novo_modelo.fit(X_train_n, y_train_num_n, sample_weight=sample_weights_n)

# 5. Avaliação do novo modelo
y_pred_num_n = xgb_novo_modelo.predict(X_test_n)
y_pred_n = le_n.inverse_transform(y_pred_num_n)

print("\n" + "="*60)
print("RELATÓRIO DE CLASSIFICAÇÃO: NOVO XGBOOST COM ENGENHARIA DE RECURSOS")
print("="*60)
print(classification_report(y_test_n, y_pred_n, target_names=class_names_n, zero_division=0))
````
